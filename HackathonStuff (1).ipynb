```python
from google import genai

username = "us"

print("Welcome to Happy Coding!", username)

level = input(
    "Choose assessment level:\n"
    "1. Beginner\n"
    "2. Intermediate\n"
    "3. Advanced\n"
)

if level == "1":
    difficulty = "Beginner"
elif level == "2":
    difficulty = "Intermediate"
elif level == "3":
    difficulty = "Advanced"
else:
    print("Invalid choice. Defaulting to Beginner.")
    difficulty = "Beginner"

print(f"\nStarting {difficulty} Assessment...")

client = genai.Client(api_key="AQ.Ab8RN6KZXcListEzSHN8ctMeUQYneDSrVDYJQOnmDnvvInXUsw")

response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents=f"""
Create 20 {difficulty} coding multiple-choice questions.

Use a mixture of:
- Multiple Choice
- True or False
- Fill in the Blank

Format each question as follows:

Question Type: Multiple Choice

Question: ...
A) ...
B) ...
C) ...
D) ...
Correct Answer: X

###

Question Type: True/False

Question: ...
Correct Answer: True

###

Question Type: Fill in the Blank

Question: Complete the code:
for i in ______(5):
Correct Answer: range

###

Separate each question with ###.
"""
)

questions = response.text.split("###")

score = 0

for question_num, question in enumerate(questions, start=1):

    if not question.strip():
        continue

    print(f"\nQuestion {question_num}/20")
    print(question)

    answer = input("Your answer (A/B/C/D): ").strip().upper()

    try:
        correct_answer = (
            question.split("Correct Answer:")[1]
            .strip()[0]
            .upper()
        )

        if answer == correct_answer:
            score += 1
            print("✅ Correct!")
        else:
            print(f"❌ Incorrect. Correct answer: {correct_answer}")

    except Exception:
        print("⚠ Could not determine correct answer.")

print("\nAssessment Complete!")
print(f"Final Score: {score}/20")

if score <= 7:
    result_level = "Beginner"
elif score <= 14:
    result_level = "Intermediate"
else:
    result_level = "Advanced"

with open("results.txt", "a") as file:
    file.write(f"{username}, {score}/20, {result_level}\n")

print("Results saved!")
print("Your coding level:", result_level)
```

```python
import matplotlib.pyplot as plt

scores = []

with open("results.txt", "r") as file:
    for line in file:
        data = line.strip().split(",")

        score = int(data[1].strip().split("/")[0])
        scores.append(score)

attempts = range(1, len(scores) + 1)

plt.plot(attempts, scores, marker="o")
plt.xlabel("Assessment Number")
plt.ylabel("Score")
plt.title("Coding Assessment Progress")
plt.grid(True)

plt.show()

if len(scores) >= 2:
    old_score = scores[-2]
    new_score = scores[-1]

    improvement = ((new_score - old_score) / old_score) * 100

    print(f"Your previous score was: {old_score}")
    print(f"Your new Score is: {new_score}")
    print(f"You have improved by: {improvement:.1f}%")
    print("Well done for continuing your practice!")
else:
    print("Not enough data to calculate improvement, sorry :(")
```
